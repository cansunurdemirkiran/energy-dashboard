# 03 — Storytelling: The Global Energy Transition
**Dataset:** owid-energy-cleaned.csv
**Purpose:** Tell a compelling, data-driven narrative about how the world is shifting away from fossil fuels — who is leading, who is lagging, and what drives the transition.

---
### Narrative Arc
The world faces an urgent challenge: decarbonize energy production fast enough to limit catastrophic climate change. This story follows the data from 1990 to 2022, tracing the slow start, the inflection point, and the countries and technologies shaping our energy future.

---
**Chapters:**
1. The Big Picture — A World in Transition
2. The Fossil Fuel Grip
3. Solar & Wind — The Exponential Story
4. Who Is Leading, Who Is Lagging?
5. The Wealth–Energy Paradox

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
import warnings, os

warnings.filterwarnings('ignore')
sns.set_style('whitegrid')
plt.rcParams.update({'figure.dpi': 110, 'font.family': 'DejaVu Sans', 'font.size': 11})

DATA_CLEAN = os.path.join('..', 'data', 'owid-energy-cleaned.csv')
df = pd.read_csv(DATA_CLEAN)
df_countries = df[df['country'] != 'World'].copy()
df_world = df[df['country'] == 'World'].sort_values('year').copy()
print('Data loaded.')
```

---
# Chapter 1: "The Big Picture — A World in Transition"

The world is getting greener, but not fast enough. The global average renewable share has crept upward since 1990, with the pace accelerating notably after the 2015 Paris Agreement mobilized international political will. Yet fossil fuels still dominate, and the gap between what's happening and what's needed remains vast.

The chart below tracks global renewable energy share from 1990 to 2022, annotated with the key political milestones that shaped energy policy.

```python
fig, ax = plt.subplots(figsize=(13, 5))

ax.plot(df_world['year'], df_world['renewables_share_energy'],
        color='#27ae60', linewidth=3, zorder=3)
ax.fill_between(df_world['year'], df_world['renewables_share_energy'],
                alpha=0.12, color='#27ae60')

# Annotate key events
events = [
    (1997, 'Kyoto Protocol'),
    (2015, 'Paris Agreement'),
    (2020, 'COVID-19 Dip'),
]
for year, label in events:
    y_val = df_world[df_world['year'] == year]['renewables_share_energy'].values
    if len(y_val) > 0:
        ax.axvline(year, color='#e74c3c', linestyle='--', alpha=0.5, linewidth=1.5)
        ax.annotate(label, xy=(year, y_val[0]), xytext=(year+0.3, y_val[0]+1.5),
                    fontsize=9.5, color='#c0392b', fontweight='bold',
                    arrowprops=dict(arrowstyle='->', color='#c0392b', lw=1))

ax.set_title('Global Renewable Energy Share: 1990–2022', fontsize=15, fontweight='bold')
ax.set_xlabel('Year', fontsize=12)
ax.set_ylabel('Renewables Share (% of total energy)', fontsize=12)
ax.set_xlim(1990, 2022)
ax.grid(True, alpha=0.3)
sns.despine()
plt.tight_layout()
plt.savefig('../assets/story_ch1.png', bbox_inches='tight')
plt.show()
```

### Chapter 1 Findings
- **The trend is real but slow:** Global renewables share has grown consistently, but the annual rate of change averages only ~0.5 percentage points per year.
- **Paris catalyzed acceleration:** Growth visibly steepened after 2015, reflecting increased policy commitment and falling technology costs.
- **COVID created a temporary dip** in 2020 as industrial activity fell, but the underlying trend resumed immediately in 2021.

---
# Chapter 2: "The Fossil Fuel Grip"

To understand the scale of the challenge, we must look not just at renewables growth but at the *composition* of the energy mix. For most of the period since 1990, coal, oil, and gas have collectively dominated global electricity generation. The stacked area chart below shows how slowly the fossil fuel grip has loosened — and how much work remains.

The key insight is that absolute renewable growth is partly offset by rising total energy demand, particularly in Asia. The *share* is growing, but total fossil fuel consumption is only beginning to plateau.

```python
stacked_cols = ['coal_share_energy', 'gas_share_energy', 'oil_share_energy',
                'hydro_share_energy', 'wind_share_energy', 'solar_share_energy']
stacked_labels = ['Coal', 'Gas', 'Oil', 'Hydro', 'Wind', 'Solar']
stacked_colors = ['#2c3e50', '#7f8c8d', '#95a5a6', '#2980b9', '#27ae60', '#f39c12']

world_stacked = df_world[['year'] + stacked_cols].set_index('year')
world_stacked.columns = stacked_labels

fig, ax = plt.subplots(figsize=(13, 6))
ax.stackplot(world_stacked.index, world_stacked.T, labels=stacked_labels,
             colors=stacked_colors, alpha=0.85)

ax.axvline(2015, color='white', linestyle='--', alpha=0.7, linewidth=1.5)
ax.text(2015.3, 95, 'Paris\nAgreement', color='white', fontsize=9, fontweight='bold')

ax.set_xlim(1990, 2022)
ax.set_ylim(0, 100)
ax.set_title('Global Energy Mix Composition (1990–2022)', fontsize=14, fontweight='bold')
ax.set_xlabel('Year', fontsize=12)
ax.set_ylabel('Share of Total Energy (%)', fontsize=12)
ax.legend(loc='lower left', framealpha=0.9, ncol=3)
sns.despine()
plt.tight_layout()
plt.savefig('../assets/story_ch2.png', bbox_inches='tight')
plt.show()
```

### Chapter 2 Findings
- **Fossil fuels collectively still account for over 60% of global energy** even in 2022 — the transition is real but far from complete.
- **Coal's share is gradually declining** in many regions, but is offset by growth in developing Asian economies.
- **Solar and wind are the new entrants** — they barely register before 2010 but show a steep upward wedge from 2015 onward.

---
# Chapter 3: "Solar & Wind — The Exponential Story"

Among all renewable sources, solar and wind are the real game-changers. Unlike hydro, which is constrained by geography and has largely been built out solar and wind are scalable, modular, and have experienced dramatic cost reductions over the past 15 years. Since 2010, the combined output of solar and wind has grown exponentially, driven by falling panel prices, improved turbine technology, and increasingly favorable policy frameworks worldwide.

```python
fig, ax1 = plt.subplots(figsize=(13, 6))
ax2 = ax1.twinx()

ax1.fill_between(df_world['year'], df_world['solar_electricity'], alpha=0.5, color='#f39c12', label='Solar (TWh)')
ax1.fill_between(df_world['year'], df_world['wind_electricity'], alpha=0.5, color='#27ae60', label='Wind (TWh)')
ax1.plot(df_world['year'], df_world['solar_electricity'], color='#d68910', linewidth=2)
ax1.plot(df_world['year'], df_world['wind_electricity'], color='#1e8449', linewidth=2)

combined_share = df_world['solar_share_energy'] + df_world['wind_share_energy']
ax2.plot(df_world['year'], combined_share, color='#8e44ad', linewidth=2.5, linestyle='--',
         marker='o', markersize=3, label='Solar+Wind Share (%)')

ax1.set_xlabel('Year', fontsize=12)
ax1.set_ylabel('Electricity Generation (TWh)', fontsize=12, color='#555')
ax2.set_ylabel('Combined Share of Energy (%)', fontsize=12, color='#8e44ad')
ax2.tick_params(axis='y', labelcolor='#8e44ad')

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left', framealpha=0.9)

ax1.set_title('Solar & Wind: From Niche to Mainstream (1990–2022)', fontsize=14, fontweight='bold')
ax1.set_xlim(1990, 2022)
ax1.grid(True, alpha=0.3)
sns.despine(right=False)
plt.tight_layout()
plt.savefig('../assets/story_ch3.png', bbox_inches='tight')
plt.show()
```

### Chapter 3 Findings
- **Solar and wind were negligible before 2005**, contributing less than 1% of global energy.
- **Post-2010 growth is exponential**, with solar doubling approximately every 3 years.
- **The inflection point coincides with grid parity** — around 2013–2015, solar became cost-competitive with fossil fuels in many markets, triggering explosive deployment.

---
# Chapter 4: "Who Is Leading, Who Is Lagging?"

Behind the global average lies enormous variation. A handful of countries, mostly in Northern Europe and South America have already built energy systems that are predominantly clean. Meanwhile, major economies in the Middle East, Central Asia, and parts of Asia remain almost entirely dependent on fossil fuels. Examining the same countries across 2000, 2010, and 2022 reveals not just where countries stand, but how quickly (or slowly) they're moving.

```python
leaders  = ['Norway', 'Brazil', 'Denmark', 'New Zealand', 'Canada']
laggers  = ['Qatar', 'Kuwait', 'Saudi Arabia', 'Iran', 'South Africa']
years_3  = [2000, 2010, 2022]

def get_country_year(countries, years):
    rows = []
    for c in countries:
        for y in years:
            val = df_countries[(df_countries['country']==c) & (df_countries['year']==y)]['renewables_share_energy']
            rows.append({'country': c, 'year': y, 'renew': val.values[0] if len(val) else np.nan})
    return pd.DataFrame(rows)

lead_data = get_country_year(leaders, years_3)
lag_data  = get_country_year(laggers, years_3)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

bar_colors = ['#a9dfbf','#52be80','#1e8449']

for ax, data, title in [
    (ax1, lead_data, '🏆 Leaders: Renewable Share Over Time'),
    (ax2, lag_data,  '⚠️  Laggers: Renewable Share Over Time')
]:
    pivot = data.pivot(index='country', columns='year', values='renew')
    x = np.arange(len(pivot))
    width = 0.25
    for i, (yr, color) in enumerate(zip(years_3, bar_colors)):
        if yr in pivot.columns:
            ax.bar(x + i*width, pivot[yr], width, label=str(yr), color=color, edgecolor='white')
    ax.set_xticks(x + width)
    ax.set_xticklabels(pivot.index, rotation=25, ha='right')
    ax.set_ylabel('Renewables Share (%)')
    ax.set_title(title, fontsize=12, fontweight='bold')
    ax.legend(title='Year')
    ax.grid(axis='y', alpha=0.3)
    sns.despine(ax=ax)

plt.suptitle('Leaders vs. Laggers: Energy Transition Progress 2000–2022', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('../assets/story_ch4.png', bbox_inches='tight')
plt.show()
```

### Chapter 4 Findings
- **Leaders like Denmark** have made dramatic gains, roughly doubling their renewables share between 2000 and 2022 thanks to aggressive wind policy.
- **Norway and Brazil** were already near 70–80% in 2000 (hydro-dominated) and have maintained or improved their position.

# Chapter 5: "The Wealth–Energy Paradox"

Conventional wisdom suggests that wealthier countries, with more technological capacity and capital, should be further along the energy transition. Yet the data reveals a paradox: many of the richest nations (particularly Gulf states and the US) have among the highest fossil fuel shares, while some middle-income countries (Brazil, Kenya, Ethiopia) lead in renewables. This scatter plot explores the complex relationship between wealth and energy transformation — and highlights the outliers who break the expected pattern.

```python
df_2022 = df_countries[df_countries['year'] == 2022].dropna(
    subset=['gdp_per_capita', 'renewables_share_energy', 'population']).copy()

continent_colors = {
    'Europe':        '#3498db',
    'Asia':          '#e74c3c',
    'Africa':        '#f39c12',
    'South America': '#27ae60',
    'North America': '#9b59b6',
    'Oceania':       '#1abc9c',
}

fig, ax = plt.subplots(figsize=(13, 8))

for continent, color in continent_colors.items():
    subset = df_2022[df_2022['continent'] == continent]
    scatter = ax.scatter(
        subset['gdp_per_capita'], subset['renewables_share_energy'],
        s=np.sqrt(subset['population']) / 50,
        c=color, alpha=0.75, label=continent, edgecolors='white', linewidth=0.5
    )

# Annotate outliers
outliers = ['Norway', 'Qatar', 'Brazil', 'Germany', 'Kenya', 'China', 'United States', 'Denmark']
for _, row in df_2022[df_2022['country'].isin(outliers)].iterrows():
    ax.annotate(row['country'],
                xy=(row['gdp_per_capita'], row['renewables_share_energy']),
                xytext=(8, 4), textcoords='offset points', fontsize=8.5,
                bbox=dict(boxstyle='round,pad=0.2', facecolor='white', alpha=0.7))

ax.set_xscale('log')
ax.set_xlabel('GDP per Capita (USD, log scale)', fontsize=12)
ax.set_ylabel('Renewables Share of Energy (%)', fontsize=12)
ax.set_title('The Wealth–Renewables Paradox (2022)\nBubble size = Population', fontsize=13, fontweight='bold')
ax.legend(title='Continent', bbox_to_anchor=(1.01, 1), loc='upper left')
ax.grid(True, alpha=0.3)
sns.despine()
plt.tight_layout()
plt.savefig('../assets/story_ch5.png', bbox_inches='tight')
plt.show()
```

### Chapter 5 Findings
- **No simple wealth–renewables relationship exists.** High-income countries span the full range from near 0% (Qatar, Kuwait) to near 100% (Norway, Iceland).
- **Resource endowment matters more than wealth:** Norway and Brazil's high renewables shares are driven by abundant hydro, not just policy.
- **The most intriguing outliers are emerging-market countries** (Kenya, Ethiopia) that have high renewables shares driven by geothermal and hydro — despite low GDP per capita.
- **The paradox for policymakers:** wealth is a necessary but not sufficient condition for energy transition. What matters is *political will and institutional design*.
